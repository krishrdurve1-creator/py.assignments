```python
a = int(input("Enter side a: "))
b = int(input("Enter side b: "))
c = int(input("Enter hypotenuse c: "))

if a**2 + b**2 == c**2:
    print("Right-angled triangle")
else:
    print("Not a right-angled triangle")
```

```python
students = ["Ram", "Sam", "John"]

print(students)

students.append("Raj")       # Add
students[0] = "Ravi"         # Update
students.remove("John")      # Delete

print(students)
```

```python
students = ("Ram", "Sam", "John")

print(students)

# Tuple cannot be changed directly
# So we make a new tuple

students = ("Ravi", "Sam", "Raj")

print(students)
```

```python
students = {1: "Ram", 2: "Sam", 3: "John"}

print(students)

students[4] = "Raj"       # Add
students[1] = "Ravi"      # Update
del students[3]           # Delete

print(students)
```

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))

if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

print("The largest number is:", largest)
```
